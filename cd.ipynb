```python
def fill_nan_category(category_column):
    x=cd.groupby('year')['category_column'].agg(pd.Series.mode)
    for i,j in zip(x.index,x):
        cd.loc[cd['year']==i,category_column]=cd.loc[cd['year']==i,category_column].fillna(j)
```

```python
def fill_nan_category(category_column):
    x = cd.groupby('year')[category_column].agg(
        lambda x: x.mode().iloc[0]
    )

    for i, j in zip(x.index, x):
        cd.loc[cd['year'] == i, category_column] = \
            cd.loc[cd['year'] == i, category_column].fillna(j)


fill_nan_category('selling_price')
fill_nan_category('km_driven')

cd[['selling_price', 'km_driven']].info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group data
ZX = cd.groupby(['transmission', 'km_driven'])['selling_price'].sum().reset_index()

# Seller type counts
DX = cd['seller_type'].value_counts().nlargest(10)

# Owner counts
PX = cd['owner'].value_counts().nlargest(10)

# Adjust layout
plt.subplots_adjust(
    left=0.1,
    bottom=0.1,
    right=0.95,
    top=0.9,
    wspace=0.4,
    hspace=0.4
)

# Graph 1
plt.subplot(2, 2, 1)

sns.barplot(
    x='km_driven',
    y='selling_price',
    hue='transmission',
    data=ZX
)

plt.title('Selling Price w.r.t Transmission and KM Driven')
plt.xticks(rotation=45)

# Graph 2
plt.subplot(2, 2, 3)

DX.plot(kind='bar')

plt.title('Seller Type')
plt.xlabel('Seller Type')
plt.ylabel('Number of Cars')
plt.xticks(rotation=45)

# Graph 3
plt.subplot(2, 2, 4)

PX.plot(kind='bar')

plt.title('Owner')
plt.xlabel('Owner')
plt.ylabel('Number of Cars')
plt.xticks(rotation=45)

plt.show()

# Display grouped data
ZX
```
